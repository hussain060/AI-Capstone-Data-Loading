```python
#Task 1: Find the dimensions of one image


from PIL import Image
import os

image_folder = "./class_0_non_agri"
image_path = os.path.join(image_folder, os.listdir(image_folder)[0])

image = Image.open(image_path)

print("Image dimensions (width, height):", image.size)
print("Image shape (height, width, channels):",
      (image.height, image.width, len(image.getbands())))
```

```python

import os
import matplotlib.pyplot as plt
from PIL import Image

dir_non_agri = "./class_0_non_agri"

image_files = sorted(os.listdir(dir_non_agri))[:4]

fig, axes = plt.subplots(1, 4, figsize=(14, 4))

for ax, filename in zip(axes, image_files):
    image = Image.open(os.path.join(dir_non_agri, filename))
    ax.imshow(image)
    ax.set_title(filename)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python

# Task 3: Create a sorted list of agricultural image paths
import os
from PIL import Image
import matplotlib.pyplot as plt

dir_agri = "./class_1_agri"

agri_images_paths = sorted([
    os.path.join(dir_agri, f)
    for f in os.listdir(dir_agri)
    if f.lower().endswith((".jpg", ".jpeg", ".png", ".tif", ".tiff"))
])

print("Task 3: First five agricultural image paths:")
print(agri_images_paths[:5])


# Task 4: Count the agricultural images
print("\nTask 4: Number of agricultural images:", len(agri_images_paths))


# Task 5: Display the first four agricultural images
fig, axes = plt.subplots(1, 4, figsize=(14, 4))

for ax, path in zip(axes, agri_images_paths[:4]):
    img = Image.open(path)
    ax.imshow(img)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
