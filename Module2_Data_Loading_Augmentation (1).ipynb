```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
import os

base_dir = "./images_dataSAT"

dir_non_agri = os.path.join(base_dir, "class_0_non_agri")
dir_agri = os.path.join(base_dir, "class_1_agri")

all_image_paths = sorted(
    [os.path.join(dir_non_agri, f) for f in os.listdir(dir_non_agri)
     if f.lower().endswith((".jpg", ".jpeg", ".png"))]
    +
    [os.path.join(dir_agri, f) for f in os.listdir(dir_agri)
     if f.lower().endswith((".jpg", ".jpeg", ".png"))]
)

print("Total image paths:", len(all_image_paths))
print("First 5 paths:", all_image_paths[:5])
```

```python
# Task 2: Create temporary list of image paths and labels
temp = list(zip(
    all_image_paths,
    [0] * 3000 + [1] * 3000
))

# Randomly select and print 5 image paths with labels
random_samples = random.sample(temp, 5)

for path, label in random_samples:
    print("Image path:", path, "| Label:", label)
```

```python
# Task 3: Generate a batch of data (batch size = 8)
train_generator = ImageDataGenerator(rescale=1./255).flow_from_directory(
    "./images_dataSAT",
    target_size=(64, 64),
    batch_size=8,
    class_mode="binary"
)

images, labels = next(train_generator)

print("Image batch shape:", images.shape)
print("Labels:", labels)
```

```python
# Task 4: Create validation data with batch size 8
validation_generator = ImageDataGenerator(rescale=1./255).flow_from_directory(
    "./images_dataSAT",
    target_size=(64, 64),
    batch_size=8,
    class_mode="binary",
    shuffle=False
)

validation_images, validation_labels = next(validation_generator)

print("Validation images shape:", validation_images.shape)
print("Validation labels:", validation_labels)
```
